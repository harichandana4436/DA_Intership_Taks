```python
text="Python is easy and python is powerful".lower()
words=text.split()
freq={}
for word in words:
    if word in freq:
        freq[word]+=1
    else:
        freq[word]=1
        
most_frequent=words[0]

for word in words:
    if freq[word] > freq[most_frequent]:
        most_frequent = word
print(most_frequent)
```
