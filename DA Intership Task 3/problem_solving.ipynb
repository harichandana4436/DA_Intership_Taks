#### next i need to split the text into word
#### next i need to check the freq so have taken empty dict
#### next i used for loop
#### next by using the if condition i have been count the word 

```python
def each_word_frequency(text):
    words=text.split()
    freq={}
    for word in words:
        if word in freq:
            freq[word]+=1
        else:
            freq[word]=1
    return freq
print(each_word_frequency("python is easy and python is powerful"))
print(each_word_frequency("hell world hello"))
```

# 4th question

## logic
#### first i need to def the function name in that i need to take s1,s2 two string
#### next i need to check the lenght of s1,s2 string if the length is not equal rteurn false
#### else i need to check the freq of s1 and s2 
#### after that if freq of s1 == freq of s2 then return True or else return False

```python
def anagrams(s1,s2):
    if len(s1)!=len(s2):
        return False
    else:
        freq1={}
        freq2={}
        for ch in s1:
            if ch in freq1:
                freq1[ch]+=1
            else:
                freq1[ch]=1
        for ch in s2:
            if ch in freq2:
                freq2[ch]+=1
            else:
                freq2[ch]=1
    if freq1 == freq2:
        return True
    else:
        return False
print(anagrams("listen","silent"))
print(anagrams("hello","world"))
```

#### by using for loop and if condition i need to check

```python
def first_repeating_character(text):
    seen=set()
    for ch in text:
        if ch in seen:
            return ch
        seen.add(ch)
    return None
print(first_repeating_character("programming"))
print(first_repeating_character("abcdef"))
```

```python
def all_zero_to_end(num):
    non_zero=[]
    for i in num:
        if i!=0:
            non_zero.append(i)
    zero_count=len(num)-len(non_zero)
    result=non_zero+[0]*zero_count

    return result
print(all_zero_to_end([0,3,0,1,2,0,4]))
print(all_zero_to_end([0,0,5,6,0,2]))
```

## logic
#### def function as remove duplicate
#### take a empty list as unique num
#### next by using for loop and if condition take if i not in unique num
#### append the i to unique number
#### and return the unique number

```python
def remove_duplicates(num):
    unique_num=[]
    for i in num:
        if i not in unique_num:
            unique_num.append(i)
    return unique_num
print(remove_duplicates([4,2,4,5,2,7,5]))
print(remove_duplicates([1,2,3,4,5,1,2,3]))
```

#### next by using for loop i need to check that i and j == target 
#### then return nums[i] and nums[j]

```python
def two_numbers_sum(num,target):
    for i in range(len(num)):
        for j in range(i+1,len(num)):
            if num[i]+num[j]==target:
                return num[i],num[j]
    return None
print(two_numbers_sum([2,7,11,15],9))
print(two_numbers_sum([3,2,4],6))
```

## logic

#### next by using for loop and if and elif condition i need to check that 
#### i < closest then closest =i
#### and i == closest and i > cosest closest=i
#### return the closest value

```python
def closest_to_zero(num):
    closest=num[0]
    for i in num:
        if abs(i)< abs(closest):
            closest = i
        elif abs(i) == abs(closest) and i > closest:
            closest = i
    return closest
print(closest_to_zero([-8,-3,2,5,-2]))
print(closest_to_zero([-4,-1,3,6]))
```

## logic
#### first def the function names as common element by l1,l1 list
#### next i nee to take a empty list to keep the common number in the empty list
#### next i need to check the element are in the two list or not
#### next if i ==j append to the two list and break the element by suing break statement

```python
def common_element(l1,l2):
    two_list_common=[]
    for i in l1:
        for j in l2:
            if i == j:
                two_list_common+=[i]
                break
    return two_list_common
print(common_element([1,2,3,4,5],[3,4,5,6,7]))
print(common_element([10,20,30],[20,30,40]))
```

# 14 th question

## logic
#### first i need to def the function name as max sum consective
#### next i need to find the max sum of two number by using num index
#### next by using for loop i need to find current sum
#### next if currecnt > max sum
#### max sum = current sum
#### return the max sum

```python
def max_sum_consective(nums):
    max_sum=nums[0]+nums[1]
    for i in range(1,len(nums)-1):
        current_sum=nums[i]+nums[i+1]

        if current_sum > max_sum:
            max_sum = current_sum
    return max_sum
print(max_sum_consective([4,7,2,9,5]))
print(max_sum_consective([1,8,9,4]))
```
